```python
import gdspy
from shapely.geometry import Polygon

# 1. Load the GDS file
lib = gdspy.GdsLibrary()
lib.read_gds('metavehicle_exact_flat.gds')

# Assume we are analyzing the top cell and a specific layer (e.g., layer 1, datatype 0)
top_cell = lib.top_level()[0]
polygons_dict = top_cell.get_polygons(by_spec=True)

target_layer = (1, 0) # (layer, datatype)
total_area = 0.0
sum_x_area = 0.0
sum_y_area = 0.0

if target_layer in polygons_dict:
    for poly_points in polygons_dict[target_layer]:
        # 2. Convert GDS polygon points to a Shapely Polygon
        shapely_poly = Polygon(poly_points)
        
        # 3. Get area and centroid from Shapely
        area = shapely_poly.area
        centroid_x = shapely_poly.centroid.x
        centroid_y = shapely_poly.centroid.y
        
        # 4. Accumulate for the weighted average
        total_area += area
        sum_x_area += (centroid_x * area)
        sum_y_area += (centroid_y * area)

# 5. Calculate final Center of Mass
if total_area > 0:
    x_cm = sum_x_area / total_area
    y_cm = sum_y_area / total_area
    print(f"Center of Mass: ({x_cm:.3f}, {y_cm:.3f})")
    print(f"Total Area: {total_area:.3f}")
else:
    print("No polygons found on the specified layer.")
```
